# Generate from your trained checkpoint
Educational example; no pretrained model quality is implied.

```python
from pathlib import Path
from llm.generation import load_model, generate
root = Path.cwd() if Path('configs').exists() else Path.cwd().parent
checkpoint = root/'demo/latest.pt'
# Run the README tiny-training commands first.
model, tokenizer, config = load_model(checkpoint)
```

```python
generate(model, tokenizer, 'Priyan learns', max_new_tokens=40, temperature=0.8)
```
