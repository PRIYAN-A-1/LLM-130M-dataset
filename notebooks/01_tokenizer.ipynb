# Train a local byte-level BPE tokenizer
Educational example; no pretrained model quality is implied.

```python
from llm.tokenizer import BPETokenizer
from llm.data import documents
from pathlib import Path
root = Path.cwd() if Path('configs').exists() else Path.cwd().parent
tok = BPETokenizer.train(documents(root/'data/raw'), root/'notebook-tokenizer.json', 512)
```

```python
text = 'Hello, தமிழ்!'
ids = tok.encode(text)
print(ids)
assert tok.decode(ids) == text
```
