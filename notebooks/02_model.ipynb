# Inspect a tiny Transformer
Educational example; no pretrained model quality is implied.

```python
import torch
from llm.config import ModelConfig
from llm.model import LanguageModel, parameter_report
c = ModelConfig(vocab_size=512, embedding_dim=128, num_heads=4, num_layers=2, mlp_dim=512, context_length=128)
model = LanguageModel(c)
parameter_report(model)
```

```python
x = torch.randint(512, (1, 32))
logits, loss = model(x[:, :-1], x[:, 1:])
loss.backward()
print(logits.shape, loss.item())
```
